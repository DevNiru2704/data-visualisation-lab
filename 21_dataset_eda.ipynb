```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("21_netflix_titles.csv")
```

```python
df
```

```python
df_editable=df.head(30).iloc[:,:10]
```

```python
df_editable
```

```python
#Bar Chart
type_counts = df['type'].value_counts()

plt.bar(type_counts.index, type_counts.values)
plt.title("Count of Content Types")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

```python
#Line Chart
year_counts = df['release_year'].value_counts().sort_index()

plt.plot(year_counts.index, year_counts.values, marker='o')
plt.title("Releases Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
#Pie Chart
plt.pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%')
plt.title("Content Distribution")
plt.show()
```

```python
#Stair Chart
plt.step(year_counts.index, year_counts.values, where='mid')
plt.title("Step Plot of Releases")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
#Histogram
plt.hist(df['release_year'], bins=10)
plt.title("Distribution of Release Years")
plt.xlabel("Year")
plt.ylabel("Frequency")
plt.show()
```

```python
#Outliers
Q1 = df['release_year'].quantile(0.25)
Q3 = df['release_year'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['release_year'] < lower_bound) | (df['release_year'] > upper_bound)]

print("Outliers:")
print(outliers)
```
